```python
import numpy as np
from scipy.stats import norm
```

```python
sample = [172,174,168,169,171,173,175,170,169,172]
population_means = 170
population_std = 3
samp_mean = np.mean(sample)
n = len(sample)
```

```python
z_score = (samp_mean - population_means) / (population_std / np.sqrt(n))
```

```python
z_score
```

```python
p_value = 2 * (1-norm.cdf(abs(z_score)))
```

```python
p_value
```

```python
alpha = 0.05

if(p_value < alpha):
    print("i will reject the null hyp")
else:
    print("i will accept the null hyp")
```
