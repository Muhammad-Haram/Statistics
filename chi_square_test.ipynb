```python
import numpy as np
import pandas as pd
import seaborn as sns
from scipy.stats import chi2_contingency
```

```python
df = sns.load_dataset('titanic');
```

```python
df
```

```python
contingency_table = pd.crosstab(df['sex'], df['survived'])
```

```python
contingency_table
```

```python
chi2,p_value,dof,expected = chi2_contingency(contingency_table)
```

```python
chi2
```

```python
p_value
```

```python
dof
```

```python
expected
```

```python
alpha = 0.05
```

```python
if p_value < alpha:
    print("I will reject the null hyp")
else:
    print('i will accept the null hyp')
```
