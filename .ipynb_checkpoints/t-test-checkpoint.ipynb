```python
import numpy as np
from scipy import stats
```

```python
sample = [172,174,168,169,171,173,175,170,169,172]
sample_mean = np.mean(sample)
sample_std = np.std(sample,ddof=1)
n = len(sample)
population_mean = 170
```

```python
t_test = (sample_mean - population_mean) / (sample_std / np.sqrt(n))
```

```python
t_test
```

```python
p_value = 2 * (1 - stats.t.cdf(abs(t_test),df = n-1))
alpha = 0.05
```

```python
p_value
```

```python
alpha = 0.05

if(p_value < alpha):
    print("i will reject the null hyp")
else:
    print("i will accept the null hyp")
```
