```python
import numpy as np
import pandas as pd
import seaborn as sns
from scipy.stats import f_oneway
```

```python
df = sns.load_dataset('titanic');
```

```python
df['pclass'].unique()
```

```python
df = df[['pclass','age']].dropna()
```

```python
class_1 = df[df['pclass'] == 1]['age']
class_2 = df[df['pclass'] == 2]['age']
class_3 = df[df['pclass'] == 3]['age']
```

```python
class_1
```

```python
class_2
```

```python
class_3
```

```python
f_test , p_value = f_oneway(class_1,class_2,class_3)
```

```python
f_test
```

```python
p_value
```

```python
alpha = 0.05
```

```python
if p_value < alpha:
    print("I will reject the null hyp")
else:
    print('i will accept the null hyp')
```
